# Sales Data Analysis
Portfolio EDA notebook.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path('..')
df = pd.read_csv(ROOT/'data/sales_data.csv', parse_dates=['date'])
df.head()
```

```python
df.info()
df.describe(numeric_only=True)
```

```python
monthly = df.assign(month=df['date'].dt.to_period('M').astype(str)).groupby('month', as_index=False)['revenue'].sum()
monthly
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=monthly, x='month', y='revenue', marker='o')
plt.xticks(rotation=45); plt.title('Monthly Revenue'); plt.tight_layout(); plt.show()
```
